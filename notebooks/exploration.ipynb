```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set style for plots
sns.set_style("whitegrid")

# 1. Define File Paths
files = {
    "amazon": "../data/raw/Amazon Sale Report.csv",
    "intl": "../data/raw/International sale Report.csv",
    "inventory": "../data/raw/Sale Report.csv",
    "expense": "../data/raw/Expense IIGF.csv",
    "pricing_may": "../data/raw/May-2022.csv",
    "warehouse": "../data/raw/Cloud Warehouse Compersion Chart.csv",
    "pricing_mar": "../data/raw/P  L March 2021.csv"
}

# 2. Load Data
dfs = {}
for name, path in files.items():
    try:
        # Expense file has a weird header structure (starts on row 2)
        if name == "expense":
            dfs[name] = pd.read_csv(path, header=1)
        else:
            # low_memory=False helps with mixed types in large files
            dfs[name] = pd.read_csv(path, low_memory=False)
        print(f"✅ Loaded {name}: {dfs[name].shape}")
    except Exception as e:
        print(f"❌ Error loading {name}: {e}")
```
